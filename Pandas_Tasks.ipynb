```python
data = {
    "Name": ["Ali", "Sara", "Ahmed"],
    "Marks": [80, None, 50]
    }
DF = pd.DataFrame(data)
#print(DF.dropna())
#print( DF["Marks"].fillna(0))
print( DF["Marks"].fillna(DF["Marks"].mean()))
```

```python
#Handling Duplicate Data

data = {
    "Name": ["Ali", "Sara", "Ali"],
    "Marks": [80, 90, 80]
}

df = pd.DataFrame(data)

#print(df.duplicated())
#print(df.duplicated().sum())
print(df.drop_duplicates())
```

```python
#Sorting Data

#print(df.sort_values("Annual Income (k$)", ascending=True))
print(df.sort_values("Annual Income (k$)", ascending=False))
```

```python
#Aggregation and GroupBy

print(df["Age"].sum())
print(df["Age"].mean())
print(df["Age"].max())
print(df["Age"].min())
print(df["Age"].count())
```

```python
#groupby()
result = df.groupby("Genre")["Spending Score (1-100)"].mean()
print(result)
```

```python
#Combining DataFrames
#concat()

df1 = pd.DataFrame({
    "Name": ["Ali", "Sara"],
    "Marks": [80, 90]
})

df2 = pd.DataFrame({
    "Name": ["Ahmed", "Ayesha"],
    "Marks": [75, 85]
})

print(pd.concat([df1 , df2],ignore_index=True))

```

```python
#merge()
students = pd.DataFrame({
    "ID": [1, 2, 3],
    "Name": ["Ali", "Sara", "Ahmed"]
})

marks = pd.DataFrame({
    "ID": [1, 2, 3],
    "Marks": [80, 90, 75]
})

print(pd.merge(students,marks,on="ID"))
```
